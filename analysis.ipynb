## Introduction

This paper studies whether financial incentives, such as matching donations, can increase charitable giving. Karlan and List (2007) conduct a large-scale field experiment in which potential donors are randomly assigned to receive or not receive a matching grant offer.

Because the assignment to treatment and control groups is random, any difference in donation behavior between the two groups can be interpreted as a causal effect of the matching incentive. This paper is important because it provides evidence on how nonprofits can design more effective fundraising strategies.

```python
import pandas as pd
import statsmodels.formula.api as smf

df = pd.read_stata("AERtables1-5.dta")
```

## Main Analysis

### Balance Check (Table 1)

The comparison of baseline characteristics between the treatment and control groups shows that they are very similar. This suggests that randomization was successful and there is no evidence of selection bias.

```python
df.groupby('treatment')[['years', 'freq']].mean().round(2)
```

### Effect of Treatment on Giving (Table 2A)

We estimate a simple linear regression to examine whether the treatment affects the probability of giving.

```python
model = smf.ols('gave ~ treatment', data=df).fit()
model.summary()
```

The regression results show that the treatment variable has a positive and statistically significant effect on donation behavior. The coefficient on treatment is approximately 0.0042, indicating that individuals in the treatment group are about 0.42 percentage points more likely to donate than those in the control group.

### Regression with Controls (Table 3)

Next, we include control variables such as prior donation history and donation frequency to test the robustness of the results.

```python
model2 = smf.ols('gave ~ treatment + years + freq', data=df).fit()
model2.summary()
```

After including control variables, the treatment effect remains positive and statistically significant. The magnitude is very similar to the baseline estimate, suggesting that the results are robust.
